# 6 Isocontours of Normal Distributions

Let $f(\mu,\Sigma)$ be the probability density function of a normally distributed random variable in $\mathbb{R}^2$.
Write code to plot the isocontours of the following functions, each on its own separate figure. Make sure it is clear which figure belongs to which part.
You're free to use any plotting libraries or stats utilities you like; for instance, in Python you can use Matplotlib and SciPy.
Choose the boundaries of the domain you plot large enough to show the interesting characteristics of the isocontours (use your judgment).
Make sure we can tell what isovalue each contour is associated with---you can do this with labels or a colorbar/legend.

```python
def normal_pdf(mu, Sigma, x1, x2):
    x = np.array([x1, x2])
    Sigma_det = np.linalg.det(Sigma)
    Sigma_inv = np.linalg.inv(Sigma)
    
    pdf = (1 / np.sqrt(2 * np.pi * Sigma_det)) * np.exp(-0.5 * (x - mu).T.dot(Sigma_inv.dot(x - mu)))
    
    return pdf
```

(1) $f(\mu,\Sigma)$, where $\mu = \begin{bmatrix} 1 \\ 1\end{bmatrix}$ and $\Sigma = \begin{bmatrix} 1 & 0\\ 0 & 2 \end{bmatrix}$.

```python
x1_i, x2_i = np.meshgrid(np.linspace(-2, 4, 100), np.linspace(-3, 5, 100))

mu = np.array([1, 1])
Sigma = np.array([[1, 0], [0, 2]])

y = np.array([normal_pdf(mu, Sigma, x1, x2) for (x1, x2) in zip(x1_i.ravel(), x2_i.ravel())]).reshape(x1_i.shape)

fig = plt.figure(figsize=(10, 4))

plt1 = fig.add_subplot(1, 2, 1)
plt1.contourf(x1_i, x2_i, y, 10, cmap='OrRd')
plt1_contour = plt1.contour(x1_i, x2_i, y, 10, linewidths=1)
plt.clabel(plt1_contour)

plt2 = fig.add_subplot(1, 2, 2, projection='3d')
plt2.plot_surface(x1_i, x2_i, y, cmap='OrRd')

plt.show();
```

(2) $f(\mu,\Sigma)$, where $\mu = \begin{bmatrix} -1 \\ 2\end{bmatrix}$ and $\Sigma = \begin{bmatrix} 2 & 1\\ 1 & 4 \end{bmatrix}$.

```python
x1_i, x2_i = np.meshgrid(np.linspace(-5, 3, 100), np.linspace(-3, 7, 100))

mu = np.array([-1, 2])
Sigma = np.array([[2, 1], [1, 4]])

y = np.array([normal_pdf(mu, Sigma, x1, x2) for (x1, x2) in zip(x1_i.ravel(), x2_i.ravel())]).reshape(x1_i.shape)

fig = plt.figure(figsize=(12, 4))

plt1 = fig.add_subplot(1, 2, 1)
plt1.contourf(x1_i, x2_i, y, 10, cmap='OrRd')
plt1_contour = plt1.contour(x1_i, x2_i, y, 10, linewidths=1)
plt.clabel(plt1_contour)

plt2 = fig.add_subplot(1, 2, 2, projection='3d')
plt2.plot_surface(x1_i, x2_i, y, cmap='OrRd')

plt.show();
```

(3) $f(\mu_1,\Sigma_1)-f(\mu_2,\Sigma_2)$, where $\mu_1 = \begin{bmatrix} 0 \\ 2\end{bmatrix}$, $\mu_2 = \begin{bmatrix} 2 \\ 0\end{bmatrix}$ and $\Sigma_1 = \Sigma_2 = \begin{bmatrix} 2 & 1\\ 1 & 1 \end{bmatrix}$.

(Yes, this is a difference between two PDFs. No, it is not itself a valid PDF. Just plot its isocontours anyway.)

```python
x1_i, x2_i = np.meshgrid(np.linspace(-3, 5, 100), np.linspace(-2, 4, 100))

mu_1 = np.array([0, 2])
mu_2 = np.array([2, 0])
Sigma = np.array([[2, 1], [1, 1]])

y_1 = np.array([normal_pdf(mu_1, Sigma, x1, x2) for (x1, x2) in zip(x1_i.ravel(), x2_i.ravel())]).reshape(x1_i.shape)
y_2 = np.array([normal_pdf(mu_2, Sigma, x1, x2) for (x1, x2) in zip(x1_i.ravel(), x2_i.ravel())]).reshape(x1_i.shape)

fig = plt.figure(figsize=(10, 4))

plt1 = fig.add_subplot(1, 2, 1)
plt1.contourf(x1_i, x2_i, y_1 - y_2, 10, cmap='OrRd')
plt1_contour = plt1.contour(x1_i, x2_i, y_1 - y_2, 10, linewidths=1)
plt.clabel(plt1_contour)

plt2 = fig.add_subplot(1, 2, 2, projection='3d')
plt2.plot_surface(x1_i, x2_i, y_1 - y_2, cmap='OrRd')

plt.show();
```

# 7 Eigenvectors of the Gaussian Covariance Matrix

```python
n = 100

mu1, mu2 = 3, 4
sigma1, sigma2 = 3, 2

X_1 = np.random.normal(mu1, sigma1, n)

X_2 = 0.5 * X_1 + np.random.normal(mu2, sigma2, n)

data = np.vstack((X_1, X_2)).T
```

(1) Compute the mean (in $\mathbb{R}^2$) of the sample.

```python
sample_mean = np.mean(data, axis=0)

print("X_1 mean:", sample_mean[0])
print("X_2 mean:", sample_mean[1])
```

(2) Compute the $2 \times 2$ covariance matrix of the sample (based on the sample mean, not the true mean---which you would not know given real-world data). Ensure that the sample covariance uses the maximum likelihood estimator as described in Question 5.

```python
cov_matrix = np.dot((data - sample_mean).T, (data - sample_mean)) / n

print(cov_matrix)
```

(3) Compute the eigenvectors and eigenvalues of this covariance matrix.

```python
eigval, eigvec = np.linalg.eig(cov_matrix)

print("Eigenvalues:\n", eigval)
print("Eigenvectors:\n", eigvec)
```

(4) On a two-dimensional grid with a horizonal axis for $X_1$ with range $[-15, 15]$ and a vertical axis for $X_2$ with range $[-15, 15]$, plot (i) all $n=100$ data points, and (ii) arrows representing both covariance eigenvectors. The eigenvector arrows should originate at the mean and have magnitudes equal to their corresponding eigenvalues.

Hint:  make sure your plotting software is set so the figure is square (i.e., the horizontal and vertical scales are the same). Not doing that may lead to hours of frustration!

```python
plt.figure(figsize=(10, 6))

plt.scatter(X_1, X_2)

for i in range(len(eigval)):
    scaled_eigvec = eigvec[:, i] * eigval[i]
    
    plt.quiver(sample_mean[0], sample_mean[1], scaled_eigvec[0], scaled_eigvec[1], scale_units='xy', scale=1, color=['r', 'b'][i])
    
    
plt.xlim(-15, 15)
plt.ylim(-15, 15)
plt.gca().set_aspect('equal')
plt.xlabel('X1')
plt.ylabel('X2')
plt.title('Data Points and Eigenvectors')
plt.grid()
plt.show();
```

(5) Let $U = [v_1~~~v_2]$ be a $2 \times 2$ matrix whose columns are the unit eigenvectors of the covariance matrix, where $v_1$ is the eigenvector with the larger eigenvalue.  We use $U^\top$ as a rotation matrix to rotate each sample point from the $(X_1, X_2)$ coordinate system to a coordinate system aligned with the eigenvectors.  (As $U^\top = U^{-1}$, the matrix $U$ reverses this rotation, moving back from the eigenvector coordinate system to the original coordinate system). Center your sample points by subtracting the mean $\mu$ from each point; then rotate each point by $U^\top$, giving $x_{\mathrm{rotated}}=U^\top(x - \mu)$.  Plot these rotated points on a new two dimensional-grid, again with both axes having range $[-15, 15]$.
(You are not required to plot the eigenvectors, which would be horizontal and vertical.)

```python
U = eigvec
x_rotated = np.dot((data - sample_mean), U)

plt.figure(figsize=(10, 6))

plt.scatter(x_rotated[:, 0], x_rotated[:, 1])

plt.xlim(-15, 15)
plt.ylim(-15, 15)
plt.gca().set_aspect('equal')
plt.xlabel('Rotated X1')
plt.ylabel('Rotated X2')
plt.title('Rotated Data Points')
plt.grid()
plt.show();
```
