```python
import numpy as np
np.random.seed(189)
```

### 1. Trace of a random matrix $A \in \mathbb{R}^{5 \times 5}$

```python
A = np.random.rand(5, 5)
einsum_trace = np.einsum('ii', A)
numpy_trace = np.trace(A)

print(np.linalg.norm(einsum_trace - numpy_trace))
```

### 2. Matrix product of random matrices $A, B \in \mathbb{R}^{5 \times 5}$

```python
A = np.random.rand(5, 5)
B = np.random.rand(5, 5)
einsum_matmul = np.einsum('ij,jk->ik', A, B)
numpy_matmul = A @ B

print(np.linalg.norm(einsum_matmul - numpy_matmul))
```

### 3. The batchwise matrix product of random matrices of shapes $(3, 4, 5)$ and $(3, 5, 6)$

```python
A = np.random.rand(3, 4, 5)
B = np.random.rand(3, 5, 6)
einsum_batch_matmul = np.einsum('ijk,ikl->ijl', A, B)
numpy_batch_matmul = A @ B

print(np.linalg.norm(einsum_batch_matmul - numpy_batch_matmul))
```
